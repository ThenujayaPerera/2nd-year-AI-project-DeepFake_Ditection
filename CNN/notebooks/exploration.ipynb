```python
import sys
sys.path.append("../src")        # so the notebook can import from src/

import matplotlib.pyplot as plt
from tensorflow import keras

from prepare_data import CLASS_NAMES, count_images, load_split
```

```python
splits = ["train", "validation", "test"]
for split in splits:
    print(split, count_images(split))

counts = [count_images(s) for s in splits]
x = range(len(splits))
plt.bar([i - 0.2 for i in x], [c["apple"] for c in counts], width=0.4, label="apple")
plt.bar([i + 0.2 for i in x], [c["tomato"] for c in counts], width=0.4, label="tomato")
plt.xticks(list(x), splits)
plt.ylabel("images")
plt.legend()
plt.show()
```

```python
train_ds = load_split("train", shuffle=True)
images, labels = next(iter(train_ds))

plt.figure(figsize=(10, 6))
for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title(CLASS_NAMES[int(labels[i][0])])
    plt.axis("off")
plt.show()
```

```python
augment = keras.Sequential([
    keras.layers.RandomFlip("horizontal"),
    keras.layers.RandomRotation(0.1),
    keras.layers.RandomZoom(0.1),
])

one_image = images[0:1]                          # keep the batch dimension
plt.figure(figsize=(8, 8))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(augment(one_image, training=True)[0].numpy().astype("uint8"))
    plt.axis("off")
plt.show()
```
